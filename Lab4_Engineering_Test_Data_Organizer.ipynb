```python
def generate_engineering_dataset(last_name, seed_num, favorite_artist):
    sources = [f"{last_name}_{t}" for t in ["T1", "T2", "T3", "T4", "T5"]]
    categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
    
    artist_clean = favorite_artist.replace(" ", "").upper()
    records = []

    for i, cat in enumerate(categories):
        source = sources[i % len(sources)]
        char_val = ord(artist_clean[i % len(artist_clean)])
        val = round(seed_num + (char_val * (i + 1) * 0.55), 2)
        records.append({"source": source, "category": cat, "value": val})
   
    extra_val1 = round(seed_num + (ord(artist_clean[0]) * 0.75), 2)
    extra_val2 = round(seed_num + (ord(artist_clean[1]) * 0.85), 2)
    records.append({"source": sources[1], "category": "TEMPERATURE", "value": extra_val1})
    records.append({"source": sources[2], "category": "VOLTAGE", "value": extra_val2})

    records.append(records[0].copy())

    organized_data = {}
    for r in records:
        cat = r["category"]
        if cat not in organized_data:
            organized_data[cat] = []
        organized_data[cat].append(r)

    seen = set()
    repeated_records = []
    distinct_records = []

    for r in records:
        item_tuple = (r["source"], r["category"], r["value"])
        if item_tuple in seen:
            repeated_records.append(r)
        else:
            seen.add(item_tuple)
            distinct_records.append(r)

    all_values = [r["value"] for r in records]
    overall_avg = sum(all_values) / len(all_values)
    min_val = min(all_values)
    max_val = max(all_values)

    cat_summaries = {}
    for cat, items in organized_data.items():
        vals = [item["value"] for item in items]
        cat_summaries[cat] = {
            "count": len(vals),
            "mean": round(sum(vals) / len(vals), 2),
            "min": min(vals),
            "max": max(vals)
        }

    ordered_results = sorted(records, key=lambda x: (x["category"], x["source"]))

    print("=" * 60)
    print("      ENGINEERING TEST DATA ORGANIZER - FINAL OUTPUT")
    print("=" * 60)

    print("\n1. GENERATED TEST DATA:")
    for i, r in enumerate(records, 1):
        print(f"   [{i}] Source: {r['source']:<12} | Category: {r['category']:<12} | Value: {r['value']}")

    print("\n2. ORGANIZED TEST DATA:")
    for cat, items in organized_data.items():
        print(f"\n   >>> Category: {cat}")
        for item in items:
            print(f"       Source: {item['source']:<12} | Value: {item['value']}")

    print("\n3. REPEATED/DISTINCT DATA:")
    print(f"   Total Records: {len(records)}")
    print(f"   Distinct Records Count: {len(distinct_records)}")
    print(f"   Repeated Records Count: {len(repeated_records)}")
    print("   Distinct Records:", distinct_records)
    print("   Repeated Entries:", repeated_records)

    print("\n4. NUMERICAL AND CATEGORICAL SUMMARIES:")
    print(f"   Overall Mean: {overall_avg:.2f} | Min: {min_val} | Max: {max_val}")
    print("   Category Breakdown:")
    for cat, stat in cat_summaries.items():
        print(f"     - {cat:<12}: Count={stat['count']}, Mean={stat['mean']}, Min={stat['min']}, Max={stat['max']}")

    print("\n5. PROCESSED RESULT:")
    for cat, items in organized_data.items():
        vals = [item["value"] for item in items]
        print(f"   Category: {cat:<12} | Count: {len(vals)} | Average Value: {round(sum(vals)/len(vals), 2)}")

    print("\n6. ORDERED RESULTS:")
    for r in ordered_results:
        print(f"   Category: {r['category']:<12} | Source: {r['source']:<12} | Value: {r['value']}")

    print("\n7. FINAL SUMMARY:")
    print(f"   Successfully processed {len(records)} test records across {len(organized_data)} categories.")
    print(f"   Found {len(distinct_records)} distinct entries and {len(repeated_records)} repeated entry.")
```
