```python
def analyze_transaction_history(last_name, seed_num, favorite_artist):
    customers = [f"{last_name}_C{i}" for i in range(1, 6)]
    product_prices = {
        "NOTEBOOK": 50,
        "PEN": 37,
        "USB": 400,
        "HEADPHONES": 1260,
        "CALCULATOR": 2000
    }
    products = list(product_prices.keys())

    artist_clean = favorite_artist.replace(" ", "").upper()
    records = []

    for i, customer in enumerate(customers):
        for j, product in enumerate(products):
            char_val = ord(artist_clean[(i + j) % len(artist_clean)])
            quantity = int(1 + ((seed_num + char_val * (j + 1)) % 5))
            price = product_prices[product]
            total_value = round(quantity * price, 2)

            records.append({
                "customer": customer,
                "product": product,
                "quantity": quantity,
                "unit_price": price,
                "transaction_value": total_value
            })

    records.append(records[0].copy())

    organized_data = {}
    for r in records: 
        cust = r["customer"]
        if cust not in organized_data:
            organized_data[cust] = []
        organized_data[cust].append(r)

    seen = set()
    repeated_transactions = []
    distinct_transactions = []

    for r in records:
        key = (r["customer"], r["product"], r["transaction_value"])
        if key in seen:
            repeated_transactions.append(r)
        else:
            seen.add(key)
            distinct_transactions.append(r)

    customer_summaries = {}
    for cust, items in organized_data.items():
        total_spent = sum(item["transaction_value"] for item in items)
        total_items = sum(item["quantity"] for item in items)
        customer_summaries[cust] = {
            "transaction_count": len(items),
            "total_items_bought": total_items,
            "total_spent": round(total_spent, 2)
        }

    product_summaries = {}
    for prod in products:
        prod_items = [r for r in records if r["product"] == prod] 
        total_qty_sold = sum(item["quantity"] for item in prod_items)
        total_revenue = sum(item["transaction_value"] for item in prod_items)
        product_summaries[prod] = {
            "times_purchased": len(prod_items),
            "total_quantity_sold": total_qty_sold,
            "total_revenue": round(total_revenue, 2)
        }

    ordered_results = sorted(records, key=lambda x: (x["customer"], x["product"]))

    overall_revenue = sum(r["transaction_value"] for r in records)
    overall_items_sold = sum(r["quantity"] for r in records)

    print("=" * 75)
    print("          TRANSACTION HISTORY ANALYZER - CONSOLIDATED REPORT")
    print("=" * 75)

    print("\n1. GENERATED TRANSACTION RECORDS:")
    for i, r in enumerate(records, 1):
        print(f"   [{i}] Customer: {r['customer']:<12} | Product: {r['product']:<12} | Qty: {r['quantity']} | Unit Price: {r['unit_price']:<6} | Total: {r['transaction_value']}")

    print("\n2. ORGANIZED TRANSACTION DATA:")
    for cust, items in organized_data.items():
        print(f"\n   >>> Customer: {cust}")
        for item in items:
            print(f"       Product: {item['product']:<12} | Qty: {item['quantity']} | Total Value: {item['transaction_value']}")

    print("\n3. REPEATED/DISTINCT TRANSACTIONS:")
    print(f"   Distinct Transactions Count: {len(distinct_transactions)}")
    print(f"   Repeated Transactions Count: {len(repeated_transactions)}")
    for r in repeated_transactions:
        print(f"     - Repeated Entry -> Customer: {r['customer']}, Product: {r['product']}, Total: {r['transaction_value']}")

    print("\n4. TRANSACTION VALUES:")
    print(f"   Total Number of Records: {len(records)}")
    print(f"   Overall Store Revenue: {overall_revenue:.2f}")

    print("\n5. CUSTOMER SUMMARIES:")
    for cust, stats in customer_summaries.items():
        print(f"   - {cust:<12}: Transactions={stats['transaction_count']}, Items Bought={stats['total_items_bought']}, Total Spent={stats['total_spent']}")

    print("\n6. PRODUCT SUMMARIES:")
    for prod, stats in product_summaries.items():
        print(f"   - {prod:<12}: Times Purchased={stats['times_purchased']}, Qty Sold={stats['total_quantity_sold']}, Revenue={stats['total_revenue']}")

    print("\n7. ORDERED RESULTS:")
    for r in ordered_results:
        print(f"   Customer: {r['customer']:<12} | Product: {r['product']:<12} | Qty: {r['quantity']} | Total: {r['transaction_value']}")

    print("\n" + "=" * 75)
    print(" 8. OVERALL TRANSACTION SUMMMARY:")
    print(f"   - Total Processed Transactions : {len(records)}")
    print(f"   - Total Units Sold             : {overall_items_sold}")
    print(f"   - Total Gross Revenue          : {overall_revenue:.2f}")
    print(f"   - Unique Customers Tracked     : {len(customers)}") 
    print("=" * 75)

    print("\n" + "=" * 75)
    print(" FINAL CONSOLIDATED TRANSACTION REPORT")
    print("=" * 75)
    print(f"   Status          : COMPLETED SUCCESSFULLY[cite: 5]")
    print(f"   Store Revenue   : {overall_revenue:.2f}")
    print(f"   Units Sold      : {overall_items_sold}")
    print(f"   Active Customers: {len(customers)}")
    print("=" * 75)    
```

```python
LAST_NAME = "Manansala"
SEED_NUM = 9
FAVORITE_ARTIST = "Joshua Garcia"

analyze_transaction_history(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
```
