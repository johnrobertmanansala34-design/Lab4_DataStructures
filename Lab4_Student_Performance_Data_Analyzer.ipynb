```python
def analyze_student_performance(last_name, seed_num, favorite_artist):
    student_labels = [f"{last_name}_S{i}" for i in range(1, 6)]
    subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

    artist_clean = favorite_artist.replace(" ", "").upper()
    records = []

    for i, student in enumerate(student_labels):
        for j, subject in enumerate(subjects):
            char_val = ord(artist_clean[(i + j) % len(artist_clean)])
            score = round(seed_num + (char_val * (j + 1) * 0.35), 2)
            score = 60 + (score % 41)

            records.append({
                "student": student,
                "subject": subject,
                "score": round(score, 2)
            })

    records.append(records[0].copy())

    records.append({
        "student": student_labels[2],
        "subject": "PHYSICS",
        "score": None
    })

    def get_category(score):
        if score is None:
            return "INCOMPLETE"
        elif score >= 90:
            return "EXCELLENT"
        elif score >= 80:
            return "GOOD"
        elif score >= 75:
            return "SATISFACTORY"
        else:
            return "PASSING"

    for r in records:
        r["category"] = get_category(r["score"])

    organized_by_student = {}
    for r in records:
        std = r["student"]
        if std not in organized_by_student:
            organized_by_student[std] = []
        organized_by_student[std].append(r)

    seen = set()
    repeated_records = []
    incomplete_records = []

    for r in records:
        if r["score"] is None:
            incomplete_records.append(r)
        else:
            key = (r["student"], r["subject"], r["score"])
            if key in seen:
                repeated_records.append(r)
            else:
                seen.add(key)

    student_summaries = {}
    for std, items in organized_by_student.items():
        valid_scores = [item["score"] for item in items if item["score"] is not None]
        avg_score = sum(valid_scores) / len(valid_scores) if valid_scores else 0
        student_summaries[std] = {
            "count": len(valid_scores),
            "average_score": round(avg_score, 2),
            "status": "Complete" if len(valid_scores) == len(subjects) else "Has Incomplete"
        }

    ordered_results = sorted([r for r in records if r["score"] is not None], key=lambda x: (x["student"], x["subject"]))

    print("=" * 70)
    print("         STUDENT PERFORMANCE DATA ANALYZER - FINAL REPORT")
    print("=" * 70)

    print("\n1. GENERATED PERFORMANCE RECORDS:")
    for i, r in enumerate(records, 1):
        score_display = r['score'] if r['score'] is not None else "MISSING"
        print(f"   [{i}] Student: {r['student']:<12} | Subject: {r['subject']:<13} | Score: {score_display}")

    print("\n2. ORGANIZED STUDENT/SUBJECT DATA:")
    for std, items in organized_by_student.items():
        print(f"\n   >>> Student: {std}")
        for item in items:
            s_val = item['score'] if item['score'] is not None else "N/A"
            print(f"       Subject: {item['subject']:<13} | Score: {s_val:<6} | Category: {item['category']}")

    print("\n3. REPEATED/INCOMPLETE RECORDS:")
    print(f"   Repeated Records Count: {len(repeated_records)}")
    for r in repeated_records:
        print(f"     - Student: {r['student']}, Subject: {r['subject']}, Score: {r['score']}")
    print(f"   Incomplete Records Count: {len(incomplete_records)}")
    for r in incomplete_records:
        print(f"     - Student: {r['student']}, Subject: {r['subject']}, Score: MISSING")

    print("\n4. PERFORMANCE CATEGORIES:")
    category_counts = {}
    for r in records:
        cat = r["category"]
        category_counts[cat] = category_counts.get(cat, 0) + 1
    for cat, count in category_counts.items():
        print(f"   - {cat:<15}: {count} records(s)")

    print("\n5. STUDENT AND SUBJECT SUMMARIES:")
    print("   Student Averages:")
    for std, stats in student_summaries.items():
        print(f"     - {std:<12}: Average Score = {stats['average_score']} ({stats['status']})")

    print("\n6. ORDERED RESULTS")
    for r in ordered_results:
        print(f"   Student: {r['student']:<12} | Subject: {r['subject']:<13} | Score: {r['score']:<6} | Category: {r['category']}")

    print("\n" + "=" * 70)
    print("    FINAL STUDENT PERFORMANCE REPORT: COMPLETED SUCCESSFULLY")
    print("=" * 70)
```
